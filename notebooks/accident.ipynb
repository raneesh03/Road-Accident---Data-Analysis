```python
df = df.drop_duplicates()
```

```python
df["Accident Date"] = pd.to_datetime(
    df["Accident Date"],errors="coerce"
)
```

```python
df.info()
```

```python
severity_count = df["Accident_Severity"].value_counts()
print(severity_count)
```

```python
severity_count.plot(kind="bar",figsize=(8,5))

plt.title("Accidents by Severity")
plt.xlabel("Accident Severity")
plt.ylabel("number of Accidents")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
df["Year"] = df["Accident Date"].dt.year
df["Month"] = df["Accident Date"].dt.month

yearly_accidents = df.groupby("Year").size()

print(yearly_accidents)
```

```python
yearly_accidents.plot(
    kind="line",
    marker="o",
    figsize=(10, 5)
)

plt.title("Yearly Road Accident Trend")
plt.xlabel("Year")
plt.ylabel("Number of Accidents")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
monthly_accidents = df.groupby("Month").size()

monthly_accidents.plot(kind="bar", figsize=(8, 5))

plt.title("Accidents by Month")
plt.xlabel("Month")
plt.ylabel("Number of Accidents")
plt.tight_layout()
plt.show()
```

```python
print("Total casualties:",
      df["Number_of_Casualties"].sum())

print("Average casualties per record:",
      df["Number_of_Casualties"].mean())

print("Maximum casualties in one record:",
      df["Number_of_Casualties"].max())
```

```python
casualties_by_severity = df.groupby(
    "Accident_Severity"
)["Number_of_Casualties"].sum()

print(casualties_by_severity)

casualties_by_severity.plot(kind="bar", figsize=(8, 5))

plt.title("Casualties by Accident Severity")
plt.xlabel("Accident Severity")
plt.ylabel("Total Casualties")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
df["Road_Surface_Conditions"].value_counts().plot(
    kind="bar", figsize=(10, 5)
)

plt.title("Accidents by Road Surface Conditions")
plt.xlabel("Road Surface")
plt.ylabel("Number of Accidents")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
df["Weather_Conditions"].value_counts().plot(
    kind="bar", figsize=(12, 5)
)

plt.title("Accidents by Weather Conditions")
plt.xlabel("Weather")
plt.ylabel("Number of Accidents")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
area_count = df["Urban_or_Rural_Area"].value_counts()

print(area_count)

area_count.plot(kind="bar", figsize=(7, 5))

plt.title("Urban vs Rural Accidents")
plt.xlabel("Area Type")
plt.ylabel("Number of Accidents")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
vehicle_count = df["Vehicle_Type"].value_counts()

print(vehicle_count.head(10))

vehicle_count.head(10).plot(
    kind="bar", figsize=(10, 5)
)

plt.title("Top 10 Vehicle Types Involved in Accidents")
plt.xlabel("Vehicle Type")
plt.ylabel("Number of Records")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
road_count = df["Road_Type"].value_counts()

print(road_count)

road_count.plot(kind="bar", figsize=(10, 5))

plt.title("Accidents by Road Type")
plt.xlabel("Road Type")
plt.ylabel("Number of Accidents")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
district_count = df["District Area"].value_counts()

print(district_count.head(10))

district_count.head(10).plot(
    kind="bar", figsize=(10, 5)
)

plt.title("Top 10 Districts by Accident Records")
plt.xlabel("District")
plt.ylabel("Number of Records")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
df.to_csv("cleaned_accident_data.csv", index=False)

print("Cleaned dataset saved successfully")
```
